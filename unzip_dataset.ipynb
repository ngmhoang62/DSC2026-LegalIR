# Google Colab Dataset Unzipper (Extract Here)

Use this small notebook to unzip `public_test_dataset.zip` directly in Google Drive.

```python
# Cell 1: Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
# Cell 2: Extract Zip File directly in place ('Extract Here')
import zipfile
import os

# Path to your uploaded zip file on Drive
zip_path = '/content/drive/MyDrive/DSC2026/LegalIR/public_test_dataset.zip'
target_dir = os.path.dirname(zip_path)

if os.path.exists(zip_path):
    print(f'Extracting {zip_path} to {target_dir}...')
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(target_dir)
    print('✅ Extraction completed successfully!')
else:
    print(f'❌ Zip file not found at {zip_path}. Please check the file path on Drive.')
```
